## **2. Gradio 익히기**

### (1) 일단 만들기 : gr.Interface

```python
# 함수 준비
def sentiment(in_text):
    return "긍정" if "좋다" in in_text else "부정"

# Gradio 인터페이스 정의
demo = gr.Interface(
    fn=sentiment,
    inputs="text",
    outputs="label",
    title="감정 분류 데모",
    description="간단한 감성분석 판별"
)

# 앱 실행
demo.launch()
```

```python
# 함수 준비
def sentiment(in_text):
    return "긍정" if "좋다" in in_text else "부정"

# Gradio 인터페이스 정의
demo = gr.Interface(
    fn=sentiment,
    inputs=gr.Textbox(lines=2, placeholder="문장을 입력하세요"),
    outputs=gr.Textbox(label="결과"),
    title="감정 분류 데모",
    description="간단한 감성분석 판별"
)

# 앱 실행
demo.launch()
```

### (2) 제대로 구성하기 : gr.Blocks

```python
with gr.Blocks(title="감정 분석 데모") as demo:
    gr.Markdown("## 감정 분석 데모")   # 화면에 보이는 제목

    txt = gr.Textbox(label="문장 입력")  # 입력 컴포넌트
    btn = gr.Button("분류")              # 실행 버튼
    out = gr.Label(label="결과")         # 출력 컴포넌트

    # 버튼 클릭 시 함수 연결
    btn.click(fn=sentiment, inputs=txt, outputs=out)

# 앱 실행
demo.launch()
```

### (3) Row, Column, Accordion

```python
with gr.Blocks(title="감정 분석 데모") as demo:
    gr.Markdown("## Row, Column 이해하기")

    with gr.Row():
        with gr.Column():
            txt = gr.Textbox(label="입력")
            btn = gr.Button("분류")
        out = gr.Label(label="결과")
    with gr.Accordion("고급 옵션", open=False):  # 접었다 펼 수 있는 구역
        slider = gr.Slider(0,1,step=0.1,label="Threshold")
demo.launch()
```

### (4) 다양한 꾸미기

```python
def classify(text, thr, lang):
    return f"입력: {text}\n임계값: {thr}\n언어: {lang}"

with gr.Blocks(title="Gradio 꾸미기 예제") as demo:
    gr.Markdown("## 🎨 화면 꾸미기 실습")

    with gr.Tab("옵션 예시"):
        with gr.Row():
            txt = gr.Textbox(label="문장 입력", placeholder="문장을 입력하세요")
            thr = gr.Slider(0.0, 1.0, step=0.1, value=0.5,
                            label="Threshold (슬라이더)")

        lang = gr.Radio(["한국어", "영어", "일본어"], label="언어 선택")
        btn = gr.Button("실행 🚀")
        out = gr.Textbox(label="출력 결과", lines=3)

        btn.click(classify, [txt, thr, lang], out)

    with gr.Tab("체크박스 & 드롭다운"):
        check = gr.Checkbox(label="동의합니다")
        drop = gr.Dropdown(["옵션A", "옵션B", "옵션C"], label="옵션 선택")
        out2 = gr.Textbox(label="선택 결과")

        def process(check, drop):
            return f"체크: {check}, 드롭다운: {drop}"

        btn2 = gr.Button("확인")
        btn2.click(process, [check, drop], out2)

demo.launch()
```
